```python
print("Predicted protein length:", len(protein))
print("Reference protein length:", len(reference_protein))
```

```python
matches = 0

for a, b in zip(protein, reference_protein):
    if a == b:
        matches += 1

print("Matches:", matches)
print("Total:", len(protein))
```

```python
identity = matches / len(protein) * 100

print("Sequence identity:", identity, "%")
```

```python
n_terminal = reference_protein[:30]

print(n_terminal)
```

```python
print("N-terminal length:", len(n_terminal))
```

```python
hydrophobic = set("AILMFWVY")
```

```python
hydrophobic_count = sum(
    aa in hydrophobic for aa in n_terminal
)

print("Hydrophobic amino acids:", hydrophobic_count)
print("Hydrophobic percentage:",
      hydrophobic_count / len(n_terminal) * 100)
```

```python
for aa in n_terminal:
    if aa in hydrophobic:
        print(aa, "HYDROPHOBIC")
    else:
        print(aa, "NOT HYDROPHOBIC")
```

```python
window_size = 10

hydrophobicity = []

for i in range(len(reference_protein) - window_size + 1):
    window = reference_protein[i:i + window_size]

    count = sum(aa in hydrophobic for aa in window)
    percentage = count / window_size * 100

    hydrophobicity.append(percentage)

print("Number of windows:", len(hydrophobicity))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(hydrophobicity)

plt.xlabel("Protein position")
plt.ylabel("Hydrophobicity (%)")
plt.title("Hydrophobicity Profile of RNase I")

plt.show()
```

## Final Interpretation

The analysis of the *Escherichia coli* K-12 MG1655 RNase I sequence identified a 807 bp gene encoding a 269-amino-acid protein. The nucleotide sequence has a GC content of 51.18%.

Protein sequence analysis identified 19 hydrophobic amino acids, corresponding to approximately 63.33% of the analyzed sequence. SignalP-6.0 predicted the presence of a Sec/SPI signal peptide with a cleavage site between positions 23 and 24, with a prediction probability of 0.964402.

These results suggest that RNase I contains an N-terminal signal peptide that may direct the protein through the Sec-dependent secretion pathway. Overall, this project demonstrates a basic computational workflow for analyzing a bacterial gene and its encoded protein using Python, Biopython, and SignalP-6.0.
